```python
from sklearn.model_selection import StratifiedKFold,cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.base import clone
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,accuracy_score
from xgboost import XGBClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.preprocessing import LabelEncoder
label= LabelEncoder()
y_train=label.fit_transform(y_train)
y_test=label.transform(y_test)
```

```python
model1=LogisticRegression()
model2=RandomForestClassifier(
    max_depth=5,
    n_estimators=100,
    random_state=42
)
model3=GaussianNB()
model4=SVC(
    C=1.0,
    gamma="scale",
    random_state=42,
    kernel="rbf"
)
model5=KNeighborsClassifier(n_neighbors=3)
model6=XGBClassifier(n_estimators=100,max_depth=5,random_state=42,learning_rate=0.1)
```

```python
model_name=["LogisticRegression","RandomForestClassifier","GaussianNB","SVC","KNeighborsClassifier","XGBClassifier"]
models=[model1,model2,model3,model4,model5,model6]
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
scorees=[]
for name,model in zip(model_name,models):
  scores=cross_val_score(model,X_train,y_train,cv=cv,scoring="accuracy")
  print(f"Name:{name}")
  print(f"scores:{scores}")
  print(f"mean score:{scores.mean()}")
  scorees.append(scores.mean())
print(scorees)
```

```python
best_index=np.argmax(scorees)
```
